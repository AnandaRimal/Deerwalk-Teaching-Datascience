```python
# Import required libraries for NLP preprocessing

import nltk
import string
import re

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

# Download required datasets
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
nltk.download('punkt_tab')
```

```python
# Example text for preprocessing

text = "Hello!!! I love learning NLP. It is amazing, and I am studying machine learning."

print("Original Text:")
print(text)
```

```python
# Convert all characters to lowercase
# This avoids treating 'NLP' and 'nlp' as different words

text = text.lower()

print("Lowercase Text:")
print(text)
```

```python
# Remove punctuation marks like ! , . ?

text = text.translate(str.maketrans('', '', string.punctuation))

print("Text without punctuation:")
print(text)
```

```python
# Remove unnecessary spaces from the text

text = " ".join(text.split())

print("Cleaned Text:")
print(text)
```

```python
# Tokenization means splitting text into smaller units

sentences = sent_tokenize(text)

print("Sentence Tokens:")
print(sentences)
```

```python
# Split sentence into individual words

words = word_tokenize(text)

print("Word Tokens:")
print(words)
```

```python
# Stop words are common words like 'is', 'the', 'and'
# These words usually do not add important meaning

stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]

print("After Stop Word Removal:")
print(filtered_words)
```

```python
# Stemming reduces words to their root form
# Example: learning -> learn

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("Stemmed Words:")
print(stemmed_words)
```

```python
# Lemmatization converts words to their dictionary base form
# Example: running -> run

lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("Lemmatized Words:")
print(lemmatized_words)
```

```python
# Combine processed words back into a sentence

final_text = " ".join(lemmatized_words)

print("Final Preprocessed Text:")
print(final_text)
```

```python
# Complete pipeline combining all preprocessing steps

def preprocess_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Tokenize words
    words = word_tokenize(text)

    # Remove stop words
    stop_words = set(stopwords.words('english'))
    words = [word for word in words if word not in stop_words]

    # Lemmatization
    lemmatizer = WordNetLemmatizer()
    words = [lemmatizer.lemmatize(word) for word in words]

    return words


sample = "I love studying Natural Language Processing and Machine Learning."

print(preprocess_text(sample))
```

```python
# Sentence Tokenization Example
# Sentence tokenization means splitting a paragraph into individual sentences.

import nltk
from nltk.tokenize import sent_tokenize

# Download tokenizer if not already downloaded
nltk.download('punkt')

# Example paragraph
text = """
Natural Language Processing is a field of Artificial Intelligence.
It helps computers understand human language.
Many applications like chatbots and translators use NLP.
"""

print("Original Paragraph:")
print(text)

# Perform sentence tokenization
sentences = sent_tokenize(text)

print("\nSentence Tokens:")
print(sentences)

# Print each sentence separately
print("\nEach sentence:")
for i, sentence in enumerate(sentences):
    print(f"Sentence {i+1}: {sentence}")
```
