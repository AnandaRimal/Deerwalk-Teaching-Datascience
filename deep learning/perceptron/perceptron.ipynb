```python
# Full Perceptron Example with Pandas and scikit-learn
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Perceptron
from sklearn.preprocessing import StandardScaler

```

```python
# 1️⃣ Create dataset using Pandas (AND logic)
data = pd.DataFrame({
    'x1': [0, 0, 1, 1],
    'x2': [0, 1, 0, 1],
    'y':  [0, 0, 0, 1]  # AND gate
})

```

```python
print("Dataset:")
print(data)

# 2️⃣ Split features and labels
X = data[['x1', 'x2']].values
y = data['y'].values

```

```python
# 3️⃣ Scale features (optional but good practice)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

#
```

```python
 # 4️⃣ Train Perceptron using scikit-learn
clf = Perceptron(max_iter=10, eta0=0.1, random_state=42)
clf.fit(X_scaled, y)
```

```python

# 5️⃣ Make predictions
y_pred = clf.predict(X_scaled)
print("\nPredictions:", y_pred)
print("Weights:", clf.coef_)
print("Bias:", clf.intercept_)

```

```python
# 6️⃣ Visualize decision boundary
plt.figure(figsize=(6,6))
for xi, label in zip(X_scaled, y):
    if label == 0:
        plt.scatter(xi[0], xi[1], color='red', marker='o', label='Class 0' if 'Class 0' not in plt.gca().get_legend_handles_labels()[1] else "")
    else:
        plt.scatter(xi[0], xi[1], color='blue', marker='x', label='Class 1' if 'Class 1' not in plt.gca().get_legend_handles_labels()[1] else "")
```

```python

# Decision boundary: w1*x1 + w2*x2 + b = 0 → x2 = -(w1*x1 + b)/w2
w = clf.coef_[0]
b = clf.intercept_[0]
x1_vals = np.linspace(-2, 2, 100)
x2_vals = -(w[0]*x1_vals + b)/w[1]
plt.plot(x1_vals, x2_vals, color='green', label='Decision Boundary')
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Perceptron Decision Boundary (scikit-learn)")
plt.legend()
plt.grid(True)
plt.show()
```
